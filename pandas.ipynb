INDEX


```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "shubham"],
    "age" : [21, 22, 19],
    "marks" : [90, 98, 99]
    
}
df = pd.DataFrame(data)
print(df.index)
print(df.columns)
print(df["marks"])
print(df["name"])
print(df["age"])
print(df[["name", "marks"]])
```

LOC AND ILOC

```python
import pandas as pd
data = {
    "marks" : [92, 90, 88],
    "city": ["pune", "goa", "delhi"],
    
    
}
index = ["aliza", "tazeen", "ayyan"]
df = pd.DataFrame(data , index=index)
print(df)
print(df.loc["aliza" , "city"])
print(df.index)
print(df.iloc[2 , 0])
print(df.loc["tazeen", "city"])
print(df.iloc[1, 1])
print(df.loc["ayyan", "marks"])
print(df.iloc[2, 0])
print(df.loc["aliza"])
print(df.iloc[0])
```
