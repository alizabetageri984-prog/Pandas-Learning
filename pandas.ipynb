LOC AND ILOC

```python
import pandas as pd
data = {
    "marks" : [92, 90, 88],
    "city": ["pune", "goa", "delhi"],
    
    
}
index = ["aliza", "tazeen", "ayyan"]
df = pd.DataFrame(data , index=index)
print(df)
print(df.loc["aliza" , "city"])
print(df.index)
print(df.iloc[2 , 0])
print(df.loc["tazeen", "city"])
print(df.iloc[1, 1])
print(df.loc["ayyan", "marks"])
print(df.iloc[2, 0])
print(df.loc["aliza"])
print(df.iloc[0])
df.loc[:, "city"]
df.iloc[:, 1]
```

Slicing rows and picking a column


```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc["a":"c", "name"])
print(df.iloc[:2, 1])
print(df.loc["a":"c", ["name"]])
print(df.loc[ "b" : "c" , "city" ])
print(df.iloc[1:3, 2])
print(df.loc[ "a" : "b" , "age" ])
print(df.iloc[ :2 , 1 ])
```

Conditions: Filtering Rows with loc

```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc[df["city"] == "tokyo"])
print(df.loc[df["age"] < 22, "name"])
```

combining two conditions

```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc[(df["age"] > 20) & (df["city"] == "london")])
print(df.loc[(df["age"] > 21) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) & (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] == "london")])
print(df.loc[(df["city"] == "tokyo") | (df["city"] == "london")])
print(df.loc[df["city"].isin(["tokyo", "london"])])
print(df.loc[(df["age"] < 22) & (df["city"] != "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london"), "city"])
```

sort value 

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.sort_values("age"))
print(df.sort_values("age", ascending=False))#Read it as: "Sort by the age column, biggest age first."
print(df.sort_values("name"))
print(df.sort_values(["city", "age"]))
print(df.sort_values("age").iloc[0])
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.sort_values("age").loc["c"])    # ananya (by label)
print(df.sort_values("age").iloc[0])     # ananya (by position)
print(df.sort_values(["city", "age"], ascending=[True, False]))
print(df.sort_values("age", ascending=False).loc[:, "name"])
print(df.sort_values("age", ascending=True)["name"])
```

groupby


```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.groupby("city")["age"].mean())
print(df.groupby("city")["age"].max())
print(df.groupby("city")["age"].min())
print(df.groupby("city")["age"].sum())
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.groupby("city")["age"].count())
print(df.groupby("city")["name"].count())
print(df.groupby("city")["age"].mean().sort_values(ascending=False))
print(df.groupby("city")["age"].max().sort_values(ascending=False))
print(df.groupby("city")["age"].min().sort_values(ascending=True))
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.groupby("city")["age"].agg(["min", "max"]))
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
result = df.groupby("city")["age"].agg(["min", "max"])
print(result.loc["tokyo"])
print(result.loc["london", "max"])
print(result.loc[result["max"] > 21])
```
