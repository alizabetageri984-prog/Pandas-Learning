```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc[(df["age"] > 20) & (df["city"] == "london")])
print(df.loc[(df["age"] > 21) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) & (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] == "london")])
print(df.loc[(df["city"] == "tokyo") | (df["city"] == "london")])
print(df.loc[df["city"].isin(["tokyo", "london"])])
print(df.loc[(df["age"] < 22) & (df["city"] != "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london"), "city"])
```

Math on a Column
Analogy first

A teacher wants to add 5 bonus marks to every student. She could add them one by one, but she says instead: “Everyone, add 5 to your marks!” Pandas works the same way. One instruction applies to the whole column at once, with no loop needed. Your plan calls this thinking in whole columns, not single numbers.

Pseudocode first
1. Make a table with marks
2. Add 5 bonus marks to every student, as a new column
3. Convert each mark to a percentage out of 100... 
4. Find the average mark of the column

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["aliza", "rahul", "ananya"],
    "marks": [92, 85, 78]
})

df["bonus"] = df["marks"] + 5
df["double"] = df["marks"] * 2
print(df)
print(df["marks"].mean())
df["half"] = df["marks"] / 2
print(df)
df["retest"] = [95, 80, 90]
df["gain"] = df["retest"] - df["marks"]
print(df[["name", "marks", "retest", "gain"]])
```

Analogy first

Two boxes in the teacher’s register are empty: rahul’s age and ananya’s marks. She can’t leave them blank, because an average would be wrong. You already know the tool: fillna. Filling a blank with a sensible estimate is called imputation.

Pseudocode first
1. Open the CSV
2. Make the city names lowercase
3. Remove the duplicate row
4. Fill the blank ages with the average age
5. Fill the blank marks with the average marks
6. Count the blanks to prove none are left

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,city,marks
aliza,21,new york,92
rahul,,london,85
ananya,19,tokyo,
kabir,20,london,70
kabir,20,london,70
meera,23,Tokyo,88"""

df = pd.read_csv(StringIO(csv_text))
df["city"] = df["city"].str.lower()
df = df.drop_duplicates()
df["age"] = df["age"].fillna(df["age"].mean())
df["marks"] = df["marks"].fillna(df["marks"].mean())

print(df)
print(df.isna().sum())
df = df.reset_index(drop=True)
print(df)
print(df.groupby("city")["marks"].mean())
print(df.groupby("city").size())
```

: Which City Has the Best Average Marks?
Analogy first

Question 1 gave you the average for each city. Now the teacher wants a ranking, with the best city on top. You already know this chain from earlier: make piles, ask a question, then rank the answers.

Pseudocode first
1. Make piles by city
2. Find the average marks in each pile
3. Sort the averages, biggest first

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,city,marks
aliza,21,new york,92
rahul,,london,85
ananya,19,tokyo,
kabir,20,london,70
kabir,20,london,70
meera,23,Tokyo,88"""

df = pd.read_csv(StringIO(csv_text))
df["city"] = df["city"].str.lower()
df = df.drop_duplicates()
df["age"] = df["age"].fillna(df["age"].mean())
df["marks"] = df["marks"].fillna(df["marks"].mean())

print(df)
print(df.isna().sum())
print(df.groupby("city")["marks"].mean().sort_values(ascending=False))
```
