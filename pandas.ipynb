Conditions: Filtering Rows with loc

```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc[df["city"] == "tokyo"])
print(df.loc[df["age"] < 22, "name"])
```

combining two conditions

```python
import pandas as pd
data = {
    "name" : ["aliza", "rahul", "ananya"],
    "age": [21, 22, 19],
    "city": ["new york", "london", "tokyo"]
}
df = pd.DataFrame(data , index = ["a", "b", "c"])
print(df.loc[(df["age"] > 20) & (df["city"] == "london")])
print(df.loc[(df["age"] > 21) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) & (df["city"] == "tokyo")])
print(df.loc[(df["age"] > 20) | (df["city"] == "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] == "london")])
print(df.loc[(df["city"] == "tokyo") | (df["city"] == "london")])
print(df.loc[df["city"].isin(["tokyo", "london"])])
print(df.loc[(df["age"] < 22) & (df["city"] != "tokyo")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london")])
print(df.loc[(df["age"] >= 21) & (df["city"] != "london"), "city"])
```

sort value 

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.sort_values("age"))
print(df.sort_values("age", ascending=False))#Read it as: "Sort by the age column, biggest age first."
print(df.sort_values("name"))
print(df.sort_values(["city", "age"]))
print(df.sort_values("age").iloc[0])
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.sort_values("age").loc["c"])    # ananya (by label)
print(df.sort_values("age").iloc[0])     # ananya (by position)
print(df.sort_values(["city", "age"], ascending=[True, False]))
print(df.sort_values("age", ascending=False).loc[:, "name"])
print(df.sort_values("age", ascending=True)["name"])
```

groupby


```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.groupby("city")["age"].mean())
print(df.groupby("city")["age"].max())
print(df.groupby("city")["age"].min())
print(df.groupby("city")["age"].sum())
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "age": [21, 22, 19, 20, 23],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d", "e"])
print(df.groupby("city")["age"].count())
print(df.groupby("city")["name"].count())
print(df.groupby("city")["age"].mean().sort_values(ascending=False))
print(df.groupby("city")["age"].max().sort_values(ascending=False))
print(df.groupby("city")["age"].min().sort_values(ascending=True))
```

dropna

```python
import pandas as pd
import numpy as np

data = {
    "name": ["aliza", "rahul", "ananya", "kabir"],
    "age": [21, np.nan, 19, 20],
    "city": ["new york", "london", None, "london"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d"])
print(df.dropna(subset=["age"]))
print(df.dropna(subset=["city"]))
print(df.dropna())
```

```python
import pandas as pd
import numpy as np

data = {
    "name": ["aliza", "rahul", "ananya", "kabir"],
    "age": [21, np.nan, 19, 20],
    "city": ["new york", "london", None, "london"]
}
df = pd.DataFrame(data, index=["a", "b", "c", "d"])
df["city"] = df["city"].fillna("unknown")
df["age"] = df["age"].fillna(df["age"].mean())
print(df)
print(df.isna().sum())
```

value_counts
at the end of an exam, the teacher makes a tally chart of how many students got each grade, such as "A: 3, B: 2". value_counts makes that tally chart for one column.

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data)

print(df["city"].value_counts())
```

```python
import pandas as pd

data = {
    "name": ["aliza", "rahul", "ananya", "kabir", "meera"],
    "city": ["new york", "london", "tokyo", "london", "tokyo"]
}
df = pd.DataFrame(data)
df["city"] = ["pune", "goa", "pune", "pune", "goa"]
print(df["city"].value_counts())
print(df["city"].value_counts(normalize=True))
df = pd.DataFrame({"city": ["pune", "goa", "pune", "pune", "goa", "goa"]})
print(df["city"].value_counts())

```

merge
 Make table 1 (name, city)
2. Make table 2 (name, marks)
3. Join them, matching rows that have the same name

```python
import pandas as pd

students = pd.DataFrame({
    "name": ["aliza", "rahul", "ananya"],
    "city": ["new york", "london", "tokyo"]
})

marks = pd.DataFrame({
    "name": ["aliza", "rahul", "ananya"],
    "marks": [92, 85, 78]
})

result = pd.merge(students, marks, on="name")
print(result)
```

```python
import pandas as pd

students = pd.DataFrame({
    "name": ["aliza", "rahul", "ananya"],
    "city": ["new york", "london", "tokyo"]
})
marks = pd.DataFrame({
    "name": ["ananya", "aliza", "rahul"],
    "marks": [78, 92, 85]
})
result = pd.merge(students, marks, on="name")
print(result)
```

```python
students = pd.DataFrame({
    "name": ["aliza", "rahul", "ananya"],
    "city": ["new york", "london", "tokyo"]
})

marks = pd.DataFrame({
    "name": ["aliza", "rahul"],
    "marks": [92, 85]
})
result = pd.merge(students, marks, on="name")
print(result)
print(pd.merge(students, marks, on="name", how="inner"))
print(pd.merge(students, marks, on="name", how="left"))
result = pd.merge(students, marks, on="name", how="left")
print(result.isna().sum())
print(result["marks"].fillna(0))
```
