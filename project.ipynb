```python
import pandas as pd
df = pd.read_csv("train (1).csv")
print(df["Survived"].value_counts(normalize=True))
print(df.groupby("Sex")["Survived"].mean())
print(df.groupby("Sex").size())
print(df.groupby("Pclass")["Survived"].mean())
```

```python
import pandas as pd
df = pd.read_csv("train (1).csv")
print(df.groupby("Pclass")["Survived"].mean())
print(df.groupby("Pclass")["Survived"].agg(["mean", "count"]))
```

```python
import pandas as pd
df = pd.read_csv("train (1).csv")
children = df.loc[df["Age"] < 12]
adults = df.loc[df["Age"] >= 12]

print(children["Survived"].mean())
print(adults["Survived"].mean())
print(len(children), len(adults))
print(df.groupby("Embarked")["Survived"].agg(["mean", "count"]))
```
