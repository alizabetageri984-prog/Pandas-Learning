```python
import pandas as pd
df = pd.read_csv("train (1).csv")
children = df.loc[df["Age"] < 12]
adults = df.loc[df["Age"] >= 12]

print(children["Survived"].mean())
print(adults["Survived"].mean())
print(len(children), len(adults))
print(df.groupby("Embarked")["Survived"].agg(["mean", "count"]))
```

```python
import pandas as pd

df = pd.read_csv("train (1).csv")
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df[["Age", "Embarked"]].isna().sum())
print(df["Embarked"].mode()[0])   # S
print(df["Age"].median())          # 28.0
```

```python
import matplotlib.pyplot as plt

df.groupby("Sex")["Survived"].mean().plot(kind="bar")
plt.title("Survival rate by sex")
plt.ylabel("Share who survived")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Age"].plot(kind="hist", bins=20)
plt.title("Age spread of passengers")
plt.xlabel("Age")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.groupby("Pclass")["Survived"].mean().plot(kind="bar")
plt.title("Survival rate by ticket class")
plt.xlabel("Ticket class")
plt.ylabel("Share who survived")
plt.show()
```

## Summary

**Data:** Titanic passengers from Kaggle, 891 rows and 12 columns.

**Cleaning:** I filled 177 blank ages with the median (28) and 2 blank ports with the most common port (S).
I left Cabin alone because 77% of it was empty.

**Findings:**
1. Women survived at 74% and men at 19%.
2. Survival fell with ticket class: 63% in class 1 and 24% in class 3.
3. Children under 12 survived at 57%, adults at 39%.

**Limitation:** Filling the blank ages created an artificial spike near 28 in the histogram.
