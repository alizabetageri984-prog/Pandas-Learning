```python
import pandas as pd

df = pd.read_csv("archive/IPL.csv")
```

```python
player_runs = df.groupby('batter')['runs_batter'].sum()

print(player_runs)
```

```python
player_runs = df.groupby('batter')['runs_batter'].sum()

print(player_runs)
```

```python
player_runs_sorted = player_runs.sort_values(ascending=False)

print(player_runs_sorted)
```

```python
highest_scorer = player_runs_sorted.idxmax()

highest_runs = player_runs_sorted.max()

print("Highest Run Scorer:", highest_scorer)
print("Total Runs:", highest_runs)
```

```python
top_10_players = player_runs_sorted.head(10)

print(top_10_players)
```

```python
season_runs = df.groupby('season')['runs_batter'].sum()

print(season_runs)
```

```python
best_season = season_runs.idxmax()
best_season_runs = season_runs.max()

print("Highest scoring season:", best_season)
print("Total runs:", best_season_runs)
```

```python
import matplotlib.pyplot as plt

season_runs.sort_index().plot(
    kind='bar',
    figsize=(12, 5)
)

plt.title('IPL Season-wise Total Runs')
plt.xlabel('Season')
plt.ylabel('Total Batter Runs')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
ax = season_runs.sort_index().plot(
    kind='bar',
    figsize=(12, 5)
)

ax.bar_label(ax.containers[0], padding=3, fontsize=8)

plt.title('IPL Season-wise Total Runs')
plt.xlabel('Season')
plt.ylabel('Total Batter Runs')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
